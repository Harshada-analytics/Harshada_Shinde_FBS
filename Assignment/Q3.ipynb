#### Q3. perrform the following operations on an array of mobile phones price 6999, 7500, 1199, 27899, 14999, 9999.
a. Create a 1d-array of mobile phones prices.

```python
import numpy as np

arr = np.array([6999, 7500, 1199, 27899, 14999, 9999 ])
arr
```

```python
arr.ndim   #check array dimension 
```

b. Convert this array to float type.

```python
arr = arr.astype(float)
arr
```

```python
arr.dtype        #check datatype
```

c. Append a new mobile having price of 13999 Rs. to this array

```python
arr = np.append(arr, 13999)
arr
```

d. Reverse this array of mobile having price of 13999 Rs. to this array 

```python
arr = np.flip(arr)
arr
```

e. Apply GST of 18% on mobile phones prices and update this array.

```python
arr = np.array([6999, 7500, 11999, 27899, 14999, 9999, 13999])

arr = arr * 1.18
arr
```

f. Sort the array in descending order of price 

```python
np.sort(arr)[::-1]
arr
```

g. What is the average mobile phone price.

```python
arr = np.mean(arr)
arr
```

H. What is the difference b/w maximum and minimum price 

```python
arr = np.max(arr) - np.min(arr)
arr 
```
